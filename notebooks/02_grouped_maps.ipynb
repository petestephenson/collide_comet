# 2 · Grouped maps of all cases

Python version of the grouped-figure section of `asymmetricPlots.m`: one panel per case
(reference top-left, the four asymmetric cases on the right) and the asymmetric / reference ratios.

Each `make_*_Plot.m` function is now one call to `grouped_slice_plot` or `grouped_ratio_plot`
with the name of a quantity.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

import collide_comet as cc
from collide_comet import plotting as ccp
from run_config import get_cases, make_saver

cases, DEMO = get_cases()
save = make_saver("grouped_maps", DEMO)
cases
```

```python
cc.list_quantities();
```

## One quantity, one view

```python
fig, axes, cbar = ccp.grouped_slice_plot(cases, "total_density", view="XZ")
save(fig, "total_density_XZ");
```

```python
fig, axes, cbar = ccp.grouped_ratio_plot(cases, "total_density", view="XZ")
save(fig, "total_density_ratio_XZ");
```

Colour limits, log scaling and colour map can all be overridden. The default colour map is the
`redblue` map the MATLAB figures used; any matplotlib colour map name works too.

```python
fig, axes, cbar = ccp.grouped_slice_plot(cases, "cold_density", view="XY", clim=(-2, 1.5), cmap="viridis")
save(fig, "cold_density_XY_viridis");
```

## Everything, every view

Set `QUANTITIES_TO_PLOT` / `VIEWS` to choose what is produced. Figures are closed after saving to
keep the notebook light; set `SHOW = True` to display them inline as well.

```python
QUANTITIES_TO_PLOT = [
    "total_density", "cold_density", "pe_density", "sw_density", "sw_warm_density",
    "secondary_density", "sw_secondary_cold_density", "neutral_density",
    "eii_frequency", "eii_production", "cold_electron_production",
    "temperature", "sw_temperature",
]
VIEWS = ["YZ", "XZ", "XY"]
SHOW = False

for quantity in QUANTITIES_TO_PLOT:
    for view in VIEWS:
        fig, *_ = ccp.grouped_slice_plot(cases, quantity, view=view)
        save(fig, f"{quantity}_{view}")
        if not SHOW:
            plt.close(fig)
```

```python
RATIOS_TO_PLOT = [
    "total_density", "cold_density", "pe_density", "sw_density", "secondary_density",
    "sw_secondary_cold_density", "eii_frequency", "eii_production", "sw_temperature",
]

for quantity in RATIOS_TO_PLOT:
    for view in VIEWS:
        fig, *_ = ccp.grouped_ratio_plot(cases, quantity, view=view)
        save(fig, f"{quantity}_ratio_{view}")
        if not SHOW:
            plt.close(fig)
```
